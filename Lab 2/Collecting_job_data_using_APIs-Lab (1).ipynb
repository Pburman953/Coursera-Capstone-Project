```python
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

response = requests.get(api_url)
response.raise_for_status()
data = response.json()
python_jobs = [job for job in data if "python" in job.get("Key Skills", "").lower()]

print("Total Python jobs:", len(python_jobs))
print()

for job in python_jobs[:5]:
    print("Id:         ", job["Id"])
    print("Job Title:  ", job["Job Title"])
    print("Location:   ", job["Location"])
    print("Key Skills: ", job["Key Skills"])
    print("-" * 40)
```

```python
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

import re
    
ALIASES = {"MySQL Server": "mysql"}
# data lists this skill as "MySQL"

def get_number_of_jobs_T(technology):
    response = requests.get(api_url)
    response.raise_for_status()
    data = response.json()

    search = ALIASES.get(technology, technology).lower()
    # match the term only when it isn't part of a longer word, or of C#/C++
    pattern = re.compile(r"(?<![\w+#])" + re.escape(search) + r"(?![\w+#])")

    number_of_jobs = sum(1 for job in data
                         if pattern.search(job.get("Key Skills", "").lower()))
    return technology, number_of_jobs
```

Calling the function for Python and checking if it works.


```python
get_number_of_jobs_T("Python")
```

#### Write a function to find number of jobs in US for a location of your choice


```python
 def get_number_of_jobs_L(location):
    response = requests.get(api_url)
    response.raise_for_status()
    data = response.json()

    number_of_jobs = 0
    for job in data:
        if location.lower() in job.get("Location", "").lower():
            number_of_jobs += 1

    print(f"Number of jobs in {location}: {number_of_jobs}")
    return location, number_of_jobs
```

Call the function for Los Angeles and check if it is working.


```python
get_number_of_jobs_L("Los Angeles")
```
